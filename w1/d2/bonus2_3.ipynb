```python
import json
from drain3 import TemplateMiner
from drain3.template_miner_config import TemplateMinerConfig

plain_text_log = "2026-06-02 16:20:00 INFO User admin requested RAG model processing in 450ms"
json_log = '{"timestamp": "2026-06-02 16:20:00", "level": "INFO", "event": "model_processing", "user": "admin", "duration_ms": 450}'

config = TemplateMinerConfig()
miner = TemplateMiner(config=config)

result_text = miner.add_log_message(plain_text_log)
print("=== PLAIN TEXT LOG ===")
print(f"Log gốc:  {plain_text_log}")
print(f"Template: {result_text['template_mined']}")

result_json = miner.add_log_message(json_log)
print("\n=== JSON LOG ĐƯA VÀO DRAIN3 ===")
print(f"Log gốc:  {json_log}")
print(f"Template: {result_json['template_mined']}")

print("\n=== CÁCH XỬ LÝ CHUẨN VỚI JSON ===")
parsed_json = json.loads(json_log)
print(f"Event: {parsed_json['event']} | User: {parsed_json['user']} | Duration: {parsed_json['duration_ms']}ms")
```

```python
import re
from drain3 import TemplateMiner
from drain3.template_miner_config import TemplateMinerConfig

nginx_log = '192.168.1.10 - - [02/Jun/2026:16:23:01 +0700] "GET /api/v1/document HTTP/1.1" 200 1024'

print("=== 1. PARSE BẰNG REGEX (MANUAL) ===")

regex_pattern = r'(?P<ip>\S+) \S+ \S+ \[(?P<time>.*?)\] "(?P<method>\S+) (?P<url>\S+) (?P<protocol>[^"]+)" (?P<status>\d{3}) (?P<size>\S+)'

match = re.match(regex_pattern, nginx_log)
if match:
    for key, value in match.groupdict().items():
        print(f" - {key.upper()}: {value}")
else:
    print("Regex không khớp!")

print("\n=== 2. PARSE BẰNG DRAIN3 (AUTO) ===")
config = TemplateMinerConfig()
miner = TemplateMiner(config=config)
result = miner.add_log_message(nginx_log)

print(f"Template sinh ra: {result['template_mined']}")
```
